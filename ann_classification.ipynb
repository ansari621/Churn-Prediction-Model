```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
import pickle

```

```python
df = pd.read_csv('Churn_Modelling.csv')
```

```python
df
```

```python
# Preprocessing the data
df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
df
```

```python
# Encoding categorical variables
label_encoder_gender = LabelEncoder()
df['Gender']= label_encoder_gender.fit_transform(df['Gender'])
df
```

```python
df['Geography'].unique()
```

```python
# One Hot Encoding for Geography column
from sklearn.preprocessing import OneHotEncoder
ohe_geo = OneHotEncoder(handle_unknown = 'ignore')
df_geo = ohe_geo.fit_transform(df[['Geography']]).toarray()
geo_df = pd.DataFrame(df_geo, columns=ohe_geo.get_feature_names_out(['Geography']))
df = pd.concat([df.drop(['Geography'], axis=1), geo_df], axis=1)
df
```

```python
# Split the dataset into features and target variable
X = df.drop(columns=['Exited'])
Y = df['Exited']
X
```

```python
Y
```

```python
# train-test split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=0, stratify=Y)
```

```python
# apply feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Save the preprocessed data to a pickle file
with open('label_encoder_gender.pkl', 'wb') as f:
    pickle.dump(label_encoder_gender, f)

with open('ohe_geo.pkl', 'wb') as f:
    pickle.dump(ohe_geo, f)

with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)
```

```python
# Train the ANN model
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
## Building the ANN model
model = Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)),
    Dense(32,activation='relu'),
    Dense(1, activation='sigmoid')

])

# compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime

## Set up the TensorBoard 
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
## set up early stopping
early_stopping_callback = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
# Train the ANN model
history = model.fit(
    X_train, Y_train, 
    validation_data=[X_test, Y_test], 
    epochs=100, 
    callbacks=[early_stopping_callback, tensorboard_callback]
    )
```
